```python
import numpy as np
import os
import matplotlib.pyplot as plt
from plyfile import PlyData
```

```python
PATH = os.path.join(os.getcwd(), "..", "data", "raw","watyaichaimonkol05.ply")
```

```python
plydata = PlyData.read(PATH)
```

```python
x = plydata['vertex']['x']
y = plydata['vertex']['y']
z = plydata['vertex']['z']
intensity = plydata['vertex']['scalar_Intensity']
```

```python
total_points = len(x)
sample_size = 100000
```

```python
indices = np.random.choice(total_points, sample_size, replace=False)
```

```python
x_sample = x[indices]
y_sample = y[indices]
z_sample = z[indices]
intensity_sample = intensity[indices]
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(x_sample, y_sample, z_sample, 
                     c=intensity_sample, cmap='viridis', 
                     s=0.5, alpha=0.8)

# ใส่แถบอธิบายสี (Colorbar)
cbar = plt.colorbar(scatter, ax=ax, shrink=0.5, aspect=5, pad=0.1)
cbar.set_label('Intensity')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.title('Point Cloud 3D Plot colored by Intensity')

plt.show()
```
