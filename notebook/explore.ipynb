```python
import numpy as np
import pandas as pd
import open3d as o3d
import math
import os
import sys
import warnings
import matplotlib.pyplot as plt
from plyfile import PlyData
warnings.filterwarnings('ignore')

from pathlib import Path
```

```python
PATH = os.path.join(os.getcwd(), "..", "data", "raw","Watchai03.ply") 
```

```python
def plot_linearary(path: str):
    pcd = PlyData.read(path)
    
    try:
        linear_value = pcd['vertex']['scalar_Linearity_(0.05)']
    except:
        print("ERROR not have scalar_Linearity_(0.05)")
        return 
    
    plt.figure(figsize=(10, 6))
    counts, bins, patches = plt.hist(linear_value, bins=100, color='royalblue', alpha=0.7, log=True)
    
    plt.title('Distribution of Geometric Linearity (Radius = 0.05m)', fontsize=14, fontweight='bold')
    plt.xlabel('Linearity Value (0.0 to 1.0)', fontsize=12)
    plt.ylabel('Frequency (Log Scale)', fontsize=12)
    plt.grid(axis='y', alpha=0.3, linestyle='--')
    
    hypothetical_threshold = 0.6
    plt.axvline(x=hypothetical_threshold, color='red', linestyle='--', linewidth=2, 
                label=f'Potential Crack Threshold (> {hypothetical_threshold})')
    
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
def plot_normal_change(path: str):
    pcd = PlyData.read(path)
    
    try:
        linear_value = pcd['vertex']['scalar_Normal_change_rate_(0.05)']
    except:
        print("ERROR not have scalar_Normal_change_rate_(0.05)")
        return 
    
    plt.figure(figsize=(10, 6))
    counts, bins, patches = plt.hist(linear_value, bins=100, color='royalblue', alpha=0.7, log=True)
    
    plt.title('Distribution of Geometric Normal change rate (Radius = 0.05m)', fontsize=14, fontweight='bold')
    plt.xlabel('Normal_change_rate Value (0.0 to 1.0)', fontsize=12)
    plt.ylabel('Frequency (Log Scale)', fontsize=12)
    plt.grid(axis='y', alpha=0.3, linestyle='--')
    
    hypothetical_threshold = 0.6
    plt.axvline(x=hypothetical_threshold, color='red', linestyle='--', linewidth=2, 
                label=f'Potential Crack Threshold (> {hypothetical_threshold})')
    
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_linearary(PATH)
```

```python
plot_normal_change(PATH)
```
